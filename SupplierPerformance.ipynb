```python
df.drop('id',axis=1,inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['SupplierLocationCode'] = le.fit_transform(df['SupplierLocationCode'])
df['SupplierCategory'] = le.fit_transform(df['SupplierCategory'])
df.head()
```

```python
num_cols = df.select_dtypes(include=['int64','float64']).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
df.head()
```

```python
print(df['Supplier_Distance'].unique())
print(df['Supplier_Distance'].value_counts())
print(df['SupplierValueCategory'].unique())
print(df['IsSupplierListedComp'].unique())
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['Supplier_Distance'] = le.fit_transform(df['Supplier_Distance'])
df['SupplierValueCategory'] = le.fit_transform(df['SupplierValueCategory'])
df['IsSupplierListedComp'] = le.fit_transform(df['IsSupplierListedComp'])
print(df.dtypes)
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(),annot=True,cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
x = df.drop('PerformInd',axis=1)
y = df['PerformInd']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(random_state=42)
rf.fit(x_train, y_train)
```

```python
rf_pred = rf.predict(x_test)
```

```python
comparison = pd.DataFrame({'Actual':y_test,'Predicted':y_pred})
print(comparison)
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test,rf_pred)
print("Accuracy:",accuracy)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,rf_pred)
print("Confusion Matrix:")
print(cm)
```

```python
from sklearn.metrics import classification_report
cr = classification_report(y_test,rf_pred)
print("Classification Report:")
print(cr)
```

```python
importance = pd.DataFrame({'Feature':x.columns,'Importance':rf.feature_importances_})
importance = importance.sort_values('Importance',ascending=True)
print(importance)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
sns.barplot(x='Importance',y='Feature',data=importance)
plt.title('Feature Importance')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(cm,annot=True,fmt='d',cmap='Blues')
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(x_train,y_train)
```

```python
lr_pred=lr.predict(x_test)
lr_accuracy = accuracy_score(y_test,lr_pred)
print("Accuracy:",lr_accuracy)
```

```python
lr_pred = lr.predict(x_test)
lr_accuracy = accuracy_score(y_test,lr_pred)
print("Accuracy:",lr_accuracy)
```

```python
print(classification_report(y_test,lr_pred))
```

```python
cm_lr = confusion_matrix(y_test,lr_pred)
print("Confusion Matrix:")
print(cm_lr)
#
```
